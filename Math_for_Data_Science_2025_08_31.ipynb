Compostie Function

```python
# f(x) = 10 + x^2
# g(x) = 3 + x

x = np.linspace(-10,10,1000)

f = lambda x : 10 + x**2
g = lambda x : 3 + x

fog = f(g(x))
gof = g(f(x))
plt.figure(figsize=(10, 4))

# f(x)
plt.subplot(1,2,1)
plt.plot(x, fog, color='red')
plt.title("f(g(x))")
plt.grid(True)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# g(x)
plt.subplot(1,2,2)
plt.plot(x, gof, color='blue')
plt.title("g(f(x))")
plt.grid(True)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

```

```python
# f(x) = x^3 + 1
# g(x) = sqrt(x+2)

x = np.linspace(-10,10,1000)
f = lambda x :  x**3 + 1
g = lambda x : np.sqrt(x+2)

print("gof(2) : ", g(f(2)))
print("fog(2) : ", f(g(2)))
print("gog(2) : ", g(g(2)))

plt.figure(figsize=(10,2))

plt.subplot(1,2,1)
mask = x >= -2
plt.plot(x[mask], f(g(x[mask])), color="red")
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.2)
plt.axvline(0, color='black', linewidth=1.2)

plt.subplot(1,2,2)
mask = f(x) >= -2
plt.plot(x[mask], g(f(x[mask])), color='blue')
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.2)
plt.axvline(0, color='black', linewidth=1.2)
```

Exponential functions


```python
# y = 3^x-1 + 2

x = np.linspace(-1,2,300)
y1 = 3**x
y2 = 3**(x-1) + 2

g1, = plt.plot(x,y1)
g2, = plt.plot(x,y2)
plt.legend(handles=(g1,g2), labels=(r'$y=3^x$', r'$y=3^{x-1}+2$'))
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.2)
plt.axvline(0, color='black', linewidth=1.2)
plt.show()
```

```python
# y = 2^-x-2

x = np.linspace(-1,2,300)

y1 = 2**(-x)
y2 = 2**(-x-2)

g1, = plt.plot(x, y1)
g2, = plt.plot(x, y2)
plt.legend(handles=(g1,g2), labels=(r'$2^{-x}$', r'$2^{-x-2}$'))
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.2)
plt.axvline(0, color='black', linewidth=1.2)
plt.show()
```

```python
# y = log2(x+1)-1

x1 = np.linspace(0.1, 4, 400)
y1 = np.log2(x1)

x2 = np.linspace(-0.9, 4, 500)
y2 = np.log2(x2+1)-1

g1, = plt.plot(x1,y1)
g2, = plt.plot(x2,y2)
plt.legend(handles=(g1,g2), labels=(r'$y=log_2(x)$',r'$y=log_2(x+1)-1$'))
plt.grid(True)
plt.axhline(0, color='black', linewidth=1.2)
plt.axvline(0, color='black', linewidth=1.2)
plt.show()
```

```python
data = np.random.exponential(scale=2.0, size=1000)
log_data = np.log(data+1)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(data, bins=30)
plt.title("Original Data (Exponential Distribution)")

plt.subplot(1,2,2)
plt.hist(log_data, bins=30)
plt.title("Log Transformed Data")
plt.show()
```

Code the class named 'Calculator' that outputs the result as shown.

```python
class Calculator:
  def __init__(self,a,b):
    self.a = a
    self.b = b

  def sum(self):
    return self.a + self.b

  def subtract(self):
    return self.a - self.b

  def product(self):
    return self.a * self.b

  def divide(self):
    return self.a / self.b
```

```python
calc1 = Calculator(5,3)
calc2 = Calculator(10,2)
```

```python
print(calc1.sum())
print(calc1.subtract())
print(calc1.product())
print(calc1.divide())
```

Quiz #0104

Code the class 'Aquarium' that outputs the result as shown.

```python
class Aquarium:
  def __init__(self):
    self.aq = []
    print("An Aquarium has opened!")

  def enters(self, a):
    if a not in self.aq:
      self.aq.append(a)
      print(f"{a} is a new member of our Aquarium!")
    else:
      print(f"We already have {a}.")

  def leaves(self, a):
    if a in self.aq:
      self.aq.remove(a)
    else:
      print("No such sea creature in the Aquarium!")

  def show_all(self):
    print(self.aq)
```

Quiz #0105

Assignment:
Write a function named 'palindrome' to test whether the entered word or sentence is a palindrome.

Ignore capitalization, spaces, punctuation marks, symbols, etc.
Use the properties of 'queue' and 'stack' data structures.

Palindrome examples:


mom

wow

noon

level

radar

kayak

racecar

God's dog

Madam, I'm Adam

```python
from collections import deque
def palindrome(d):
  cleaned = ''.join(ch.lower() for ch in d if ch.isalnum())
  print(cleaned)

  stack = []
  q = deque()

  for i in cleaned:
    stack.append(i)
    q.append(i)

  while stack:
    if stack.pop() != q.popleft():
      print(False)
      return
  print(True)
  return
```

```python
palindrome("Madam, I'm Adam")
```
