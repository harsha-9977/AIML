```python
"""
Colab-ready notebook: Practical Deep Learning with TensorFlow — Stock Price Prediction (T4 GPU)

How to use in Google Colab
--------------------------
1. Open a new Colab notebook.
2. Runtime -> Change runtime type -> Hardware accelerator -> GPU (select T4 if available).
3. Copy this entire notebook into a cell or upload as a .py/.ipynb file.
4. Run cells top-to-bottom. If you want to save models/data to Google Drive, run the Drive mount cell.

Notes for T4 GPU
----------------
- This notebook uses TensorFlow 2.x and enables mixed precision to speed up training on T4.
- The model's final Dense layer is forced to float32 to avoid numeric issues with mixed precision.
- Installs required packages (yfinance, ta) in the notebook (Colab ephemeral environment).

Dependencies (installed inside the notebook)
- tensorflow (preinstalled in Colab)
- yfinance
- ta
- scikit-learn
- matplotlib
- joblib

"""

# %%
# --- Colab setup and installs
# Run this cell at the top of the Colab notebook

# If running in Colab, install packages not present by default
try:
    import google.colab  # type: ignore
    IN_COLAB = True
except Exception:
    IN_COLAB = False

if IN_COLAB:
    # upgrade pip quietly and install missing packages
    !pip -q install yfinance ta joblib

# Optional: mount Google Drive to save models and outputs
# from google.colab import drive
# drive.mount('/content/drive')

# %%
# Imports and configuration
import os
import math
import datetime as dt
from typing import Tuple, List

import numpy as np
import pandas as pd

import yfinance as yf

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import matplotlib.pyplot as plt

# Colab: show GPU
print('TensorFlow version:', tf.__version__)
print('GPU available:', tf.config.list_physical_devices('GPU'))

# Enable mixed precision for speed on T4 (optional but recommended)
from tensorflow.keras import mixed_precision
policy = mixed_precision.Policy('mixed_float16')
mixed_precision.set_global_policy(policy)
print('Mixed precision policy:', mixed_precision.global_policy())

# %%
# CONFIG - change as needed
CONFIG = {
    "TICKER": "AAPL",                # change to your chosen stock
    "START_DATE": "2010-01-01",
    "END_DATE": dt.date.today().strftime("%Y-%m-%d"),
    "SEQ_LEN": 60,                    # days of history used to predict next day
    "PREDICT_FORWARD": 1,             # days ahead to predict (1 = next day)
    "TEST_SIZE": 0.1,                 # fraction for test
    "VAL_SIZE": 0.1,                  # fraction for validation (of train)
    "BATCH_SIZE": 64,
    "EPOCHS": 50,
    "RANDOM_SEED": 42,
    "MODEL_DIR": "models",
}

os.makedirs(CONFIG["MODEL_DIR"], exist_ok=True)
np.random.seed(CONFIG["RANDOM_SEED"]) ; tf.random.set_seed(CONFIG["RANDOM_SEED"])

# %%
# 1) Data download

def download_stock(ticker: str, start: str, end: str) -> pd.DataFrame:
    df = yf.download(ticker, start=start, end=end, progress=False)
    df = df.dropna()
    return df

print("Downloading data for", CONFIG["TICKER"])
raw_df = download_stock(CONFIG["TICKER"], CONFIG["START_DATE"], CONFIG["END_DATE"])
print("Rows:", raw_df.shape[0])
raw_df.head()

# %%
# Quick plot of close price
plt.figure(figsize=(12,5))
plt.plot(raw_df.index, raw_df['Close'])
plt.title(f"{CONFIG['TICKER']} Close Price")
plt.xlabel('Date')
plt.ylabel('Close')
plt.show()

# %%
# 2) Feature engineering

def add_technical_indicators(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df['SMA_7'] = df['Close'].rolling(window=7).mean()
    df['SMA_21'] = df['Close'].rolling(window=21).mean()
    df['EMA_12'] = df['Close'].ewm(span=12, adjust=False).mean()
    df['EMA_26'] = df['Close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = df['EMA_12'] - df['EMA_26']
    df['MACD_SIGNAL'] = df['MACD'].ewm(span=9, adjust=False).mean()
    df['Momentum'] = df['Close'] - df['Close'].shift(10)
    delta = df['Close'].diff()
    up = delta.clip(lower=0)
    down = -1 * delta.clip(upper=0)
    roll_up = up.rolling(14).mean()
    roll_down = down.rolling(14).mean()
    rs = roll_up / (roll_down + 1e-8)
    df['RSI'] = 100.0 - (100.0 / (1.0 + rs))
    df['Volatility_21'] = df['Close'].rolling(window=21).std()
    return df

feat_df = add_technical_indicators(raw_df)
feat_df = feat_df.dropna()
feat_df.tail()

# %%
# 3) Labeling and splits

def add_labels(df: pd.DataFrame, forward: int = 1, label_col: str = 'Close') -> pd.DataFrame:
    df = df.copy()
    df['Target'] = df[label_col].shift(-forward)
    df = df.dropna()
    return df

labeled_df = add_labels(feat_df, forward=CONFIG['PREDICT_FORWARD'])

feature_cols = [
    'Open','High','Low','Close','Volume',
    'SMA_7','SMA_21','EMA_12','EMA_26','MACD','MACD_SIGNAL',
    'Momentum','RSI','Volatility_21'
]

X_all = labeled_df[feature_cols]
y_all = labeled_df[['Target']]

n = X_all.shape[0]
test_n = int(CONFIG['TEST_SIZE'] * n)
trainval_n = n - test_n

X_trainval = X_all.iloc[:trainval_n]
y_trainval = y_all.iloc[:trainval_n]
X_test = X_all.iloc[trainval_n:]
y_test = y_all.iloc[trainval_n:]

val_n = int(CONFIG['VAL_SIZE'] * trainval_n)
train_n = trainval_n - val_n

X_train = X_trainval.iloc[:train_n]
y_train = y_trainval.iloc[:train_n]
X_val = X_trainval.iloc[train_n:]
y_val = y_trainval.iloc[train_n:]

print('Train samples:', X_train.shape[0], 'Val samples:', X_val.shape[0], 'Test samples:', X_test.shape[0])

# %%
# 4) Scaling and sequence preparation

X_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()
X_scaler.fit(X_train)
y_scaler.fit(y_train)

X_train_s = pd.DataFrame(X_scaler.transform(X_train), index=X_train.index, columns=X_train.columns)
X_val_s = pd.DataFrame(X_scaler.transform(X_val), index=X_val.index, columns=X_val.columns)
X_test_s = pd.DataFrame(X_scaler.transform(X_test), index=X_test.index, columns=X_test.columns)

y_train_s = pd.DataFrame(y_scaler.transform(y_train), index=y_train.index, columns=y_train.columns)
y_val_s = pd.DataFrame(y_scaler.transform(y_val), index=y_val.index, columns=y_val.columns)
y_test_s = pd.DataFrame(y_scaler.transform(y_test), index=y_test.index, columns=y_test.columns)

def make_sequences(X: pd.DataFrame, y: pd.DataFrame, seq_len: int) -> Tuple[np.ndarray, np.ndarray, List[pd.Timestamp]]:
    X_vals = X.values
    y_vals = y.values
    n = len(X_vals)
    seq_X, seq_y = [], []
    seq_dates = []
    for i in range(seq_len, n):
        seq_X.append(X_vals[i-seq_len:i])
        seq_y.append(y_vals[i])
        seq_dates.append(X.index[i])
    return np.array(seq_X), np.array(seq_y), seq_dates

SEQ_LEN = CONFIG['SEQ_LEN']
X_tr_seq, y_tr_seq, dates_tr = make_sequences(X_train_s, y_train_s, SEQ_LEN)
X_val_seq, y_val_seq, dates_val = make_sequences(X_val_s, y_val_s, SEQ_LEN)
X_test_seq, y_test_seq, dates_test = make_sequences(X_test_s, y_test_s, SEQ_LEN)

print('Sequence shapes:', X_tr_seq.shape, y_tr_seq.shape)

# %%
# 5) Build LSTM model optimized for mixed precision

def build_lstm_model(input_shape: Tuple[int,int], units: int = 128, dropout: float = 0.2) -> keras.Model:
    inp = keras.Input(shape=input_shape)
    x = layers.LSTM(units, return_sequences=True)(inp)
    x = layers.Dropout(dropout)(x)
    x = layers.LSTM(int(units/2))(x)
    x = layers.Dropout(dropout)(x)
    # final Dense forced to float32 to avoid dtype issues with mixed precision
    out = layers.Dense(1, dtype='float32')(x)
    model = keras.Model(inp, out)
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=1e-3), loss='mse', metrics=['mae'])
    return model

input_shape = (X_tr_seq.shape[1], X_tr_seq.shape[2])
model = build_lstm_model(input_shape, units=128, dropout=0.2)
model.summary()

# %%
# 6) Training
callbacks = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
    keras.callbacks.ModelCheckpoint(os.path.join(CONFIG['MODEL_DIR'], 'best_lstm.h5'), save_best_only=True, monitor='val_loss'),
]

history = model.fit(
    X_tr_seq, y_tr_seq,
    validation_data=(X_val_seq, y_val_seq),
    epochs=CONFIG['EPOCHS'],
    batch_size=CONFIG['BATCH_SIZE'],
    callbacks=callbacks,
    verbose=2
)

# plot training history
plt.figure(figsize=(8,4))
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend()
plt.title('Training Loss')
plt.show()

# %%
# 7) Evaluation

y_pred_s = model.predict(X_test_seq)
y_pred = y_scaler.inverse_transform(y_pred_s)
y_true = y_scaler.inverse_transform(y_test_seq)

mse = mean_squared_error(y_true, y_pred)
rmse = math.sqrt(mse)
mae = mean_absolute_error(y_true, y_pred)

print(f'Test RMSE: {rmse:.4f}  MAE: {mae:.4f}')

plt.figure(figsize=(12,6))
plt.plot(dates_test, y_true, label='True')
plt.plot(dates_test, y_pred, label='Predicted')
plt.title(f'{CONFIG["TICKER"]} Close Price Prediction')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()

# %%
# 8) Save model and scalers (optionally to Google Drive)
model.save(os.path.join(CONFIG['MODEL_DIR'], 'final_lstm.keras'))
import joblib
joblib.dump(X_scaler, os.path.join(CONFIG['MODEL_DIR'], 'X_scaler.save'))
joblib.dump(y_scaler, os.path.join(CONFIG['MODEL_DIR'], 'y_scaler.save'))
print('Saved model and scalers to', CONFIG['MODEL_DIR'])

# %%
# 9) Predict next day from most recent sequence

def predict_next(model: keras.Model, X_scaler, y_scaler, recent_df: pd.DataFrame, seq_len: int):
    Xs = X_scaler.transform(recent_df)
    seq = np.expand_dims(Xs[-seq_len:], axis=0)
    pred_s = model.predict(seq)
    pred = y_scaler.inverse_transform(pred_s)
    return float(pred[0,0])

if len(X_test_s) >= SEQ_LEN:
    next_pred = predict_next(model, X_scaler, y_scaler, X_test_s, SEQ_LEN)
    print(f'Next-day predicted price: {next_pred:.4f}')

# %%
# 10) Short notes and next steps
# - The notebook is Colab/T4-ready. For reproducible experiments across runs, save your model
#   and summary outputs to Google Drive.
# - To speed up experimentation: increase BATCH_SIZE, reduce EPOCHS, or use smaller models.
# - For production OR more robust evaluation use walk-forward cross-validation and avoid
#   leaking future information in feature engineering.

# End of notebook
```
