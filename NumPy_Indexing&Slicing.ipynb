***NumPy Slicing***

```python
b1 = np.array([0,10,20,30,40,50])
b1[1:4]
```

```python
b1[:3]
```

```python
b1[2:]
```

```python
arr2d = np.arange(1,10).reshape(3,3)
arr2d
```

```python
arr2d[:2]
```

```python
arr2d[:2,1:]
```

```python
arr2d[1,:2]
```

```python
arr2d[:2,2]
```

```python
arr2d[:,:1]
```

```python
a = np.arange(10,100,10).reshape(3,3)
a
```

```python
a[:2,1:]
```

```python
a[2]
```

```python
a[2,:]
```

```python
a[2:,:]
```

```python
a[:,:2]
```

```python
a[1,:2]
```
