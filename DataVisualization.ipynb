Bar Plot

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

x = np.array(['Q1','Q2','Q3','Q4'])
y = np.array([1,2,3,4])
plt.bar(x,y, color="purple")
plt.title("Bar Plot")
plt.show()
```

Histogram

```python
x = np.random.randn(10000)
plt.hist(x, bins=20, color='green', density=True)
plt.title('Histogram')
plt.show()
```

Box Plot


```python
x = np.random.randn(10000) * 10 + 3
y = np.random.randn(10000) * 10 + 5
z = np.random.randn(10000) * 10 + 1
plt.boxplot([x,y,z],0)
plt.title("Box Plot")
plt.show()
```

Line Plot

```python
x = np.linspace(0,10,100)
y = np.sin(x)
plt.plot(x,y,color='orange',linewidth=2)
plt.xlabel('X')
plt.ylabel('Sin')
plt.title('Line Plot')
plt.show()
```

MatplotLib Visualization with Objects

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)
```

```python
fig0 = plt.figure()
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x, y, color='red', linestyle='-')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('Line Plot')
# axes0.set_xlim(2,4)
# axes0.set_ylim(-2,2)
plt.show()
```

```python
fig0 = plt.figure(figsize=(5,2))
axes0 = fig0.add_axes([0,0,1,1])
axes0.plot(x,y, color='orange', linestyle='--', label='Sin')
axes0.plot(x,z, color='purple', linestyle='-.', label='Cos')
axes0.legend(loc=0)
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('Line Plot')
plt.show()
```

```python
fig0 = plt.figure(figsize=(8,3))
axes0 = fig0.add_axes([0,0,0.3,1])
axes1 = fig0.add_axes([0.5,0,0.3,1])
axes0.plot(x,y, color='red', linestyle=':')
axes1.plot(x,z, color='blue', linestyle='-.')
axes0.set_xlabel('X')
axes0.set_ylabel('Y')
axes0.set_title('SINE')
axes1.set_xlabel('X')
axes1.set_ylabel('Y')
axes1.set_title('COS')
plt.show()
```

Multiple plots


```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(6,6))

# 0,0
c = np.random.randn(10000)
axes[0,0].hist(c, bins=100, color='green', density=True, alpha=0.3)
axes[0,0].set_title('Histogram')

# 1,0
axes[1,0].plot(x,z, linestyle='--')
axes[1,0].set_xlabel('X')
axes[1,0].set_ylabel('Y')
axes[1,0].set_title('Cos')

# 0,1
a = np.array(['Q1','Q2','Q3','Q4'])
b = np.array([2,4,3,1])
axes[0,1].bar(a,b, color='blue', alpha=0.5)
axes[0,1].set_title('BarPlot')

# 1,1
axes[1,1].plot(x,y,color='green', linestyle=":")
axes[1,1].set_xlabel('X')
axes[1,1].set_ylabel('Y')
axes[1,1].set_title('SINE')

plt.tight_layout()
plt.show()
```
