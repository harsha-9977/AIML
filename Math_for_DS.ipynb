```python
#5 House price
# from sklearn.datasets import load_boston
# boston = load_boston()
import pandas as pd
boston = pd.read_csv("BostonHousing.csv")
boston.head()
```

```python
boston.columns
```

```python
x = boston.drop('medv',axis=1).values
y = boston['medv'].values

a = x[:4,[0,4,5,6]]
b = y[:4]
```

```python
#5
sol = np.linalg.solve(a,b,rcond=None)
sol
```

```python
features = ['CRIM','NOX','RM',"AGE"]
for name,weight in zip(features,sol):
  print(f'{name} weight: {weight:.4f}')
```

```python
#6
w, residuals, rank, s = np.linalg.lstsq(x,y,rcond=None)

feature_names = boston.columns
for name,weight in zip(feature_names,w):
  print(f'{name} : {weight:.4f}')
```

```python
#7  vector distance
from numpy.linalg import norm
a = np.array([4,5,2,2])
b = np.array([4,0,2,0])
c = np.array([2,2,0,1])

def euclidean(u,v):
  return norm(u - v)

def cosine(u,v):
  return np.dot(u,v) / (norm(u) * norm(v))

print("Euclidean")
print(f'(a,b) : {euclidean(a,b)}')
print(f'(b,c) : {euclidean(b,c)}')
print(f'(a,c) : {euclidean(a,c)}')
print("Cosine")
print(f'(a,b) : {cosine(a,b)}')
print(f'(b,c) : {cosine(b,c)}')
print(f'(a,c) : {cosine(a,c)}')
```

```python
#11
#1. f(x) = x^2-1
import sympy as sp

x = sp.Symbol('x')
fx = x**3-1
result = sp.diff(fx,x)
result
```

```python
#2. fx = log(x^2-3k)
x,k = sp.symbols('x k')
fx = sp.log(x**2 - 3*k)
result = sp.diff(fx,x)
result
```

```python
#3. fx = exp(ax^b)
x,a,b = sp.symbols('x a b')
fx = sp.exp(a * x**b)
result = sp.diff(fx,x)
result
```
