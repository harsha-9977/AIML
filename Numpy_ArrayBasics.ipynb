```python
import numpy as np
import matplotlib.pyplot as plt

# rand vs randn
plt.hist(np.random.rand(1000), bins=30, alpha=0.6, label='rand() [0-1]')
plt.hist(np.random.randn(1000), bins=30, alpha=0.6, label='randn() N(0,1)')
plt.legend()
plt.title("rand() vs randn() Distributions")
plt.show()
```

**Adding elements to numpy arrays**

```python
a = np.array([1,2,3])
```

```python
b = np.append(a,[4,5,6])
b
```

```python
a = np.array([[1,2],[3,4]])
a
```

```python
b = np.append(a,[[9,9]],axis=0)
b
```

```python
c = np.append(a,[[9],[9]],axis=1)
c
```
