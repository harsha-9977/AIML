```python
import pandas as pd
```

```python
df_left = {
    'Name' : ['Harry Potter', 'David Baker', 'John Smith', 'Jaun Martinez', 'Jane Conner'],
    'Gender' : ['Male','Male','Male','Male','Female'],
    'Age' : [23,31,22,36,30]
}
```

```python
df_left = pd.DataFrame(df_left)
df_left
```

```python
df_right = {
    'Name' : ['John Smith', 'Alex Du Bois', 'Joanne Rowling', 'Jane Conner'],
    'Position' : ['Intern', 'Team Lead', 'Manager', 'Manager'],
    'Wage' : [25000, 75000, 90000, 70000]
}
```

```python
df_right = pd.DataFrame(df_right)
df_right
```

```python
# Inner Join!
pd.merge(df_left, df_right, on='Name')
```

```python
# Inner Join!
pd.merge(df_left, df_right, left_on='Name', right_on='Name', how='inner')
```

```python
# Left Join!
pd.merge(df_left, df_right, left_on='Name', right_on='Name', how='left')
```

```python
# Right Join!
pd.merge(df_left, df_right, left_on='Name', right_on='Name', how='right')
```

```python
# Outer Join!
pd.merge(df_left, df_right, left_on='Name', right_on='Name', how='outer')
```

```python
pd.concat([df_left, df_right], sort=True)
```

```python
pd.concat([df_left, df_right], axis=1, sort=True)
```
