```python
from ultralytics import YOLO

# Load a pre-trained model (you can also use yolov8n, yolov8s, yolov8m, etc.)
model = YOLO("yolov8n.pt")  # or "yolov8s.pt" for more accuracy

# Train the model
model.train(
    data="/content/Cyldet-1/data.yaml",  # path to data.yaml
    epochs=50,
    imgsz=640,
    batch=16,
    name="yolov8_cylinders_cyldet"
)
```

```python
from ultralytics import YOLO
import cv2

# Load the trained model
model = YOLO("/content/runs/detect/yolov8_cylinders_cyldet/weights/best.pt")

# Load your video (replace with your actual path if different)
video_path = "/content/cylinder.mp4"  # ✅ Change this to your video filename
cap = cv2.VideoCapture(video_path)

# Set up video writer (optional: to save output)
output_path = "/content/output_cylinder_detection.mp4"
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
fps = int(cap.get(cv2.CAP_PROP_FPS))
w, h = int(cap.get(3)), int(cap.get(4))
out = cv2.VideoWriter(output_path, fourcc, fps, (w, h))

frame_number = 0
total_cylinder_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Predict using YOLOv8
    results = model.predict(source=frame, conf=0.4, verbose=False)

    # Draw results on frame
    annotated_frame = results[0].plot()

    # Count cylinders in current frame
    cylinder_count = len(results[0].boxes.cls.cpu().numpy())
    total_cylinder_count += cylinder_count

    # Write to output
    out.write(annotated_frame)

    # Optional: print frame stats
    print(f"Frame {frame_number}: {cylinder_count} cylinders detected")
    frame_number += 1

cap.release()
out.release()

print("✅ Inference complete!")
print("📦 Output video saved to:", output_path)
print("🧮 Total cylinder count across all frames:", total_cylinder_count)
```

```python
from ultralytics import YOLO
import cv2

# Load trained model
model = YOLO("/content/runs/detect/yolov8_cylinders_cyldet/weights/best.pt")

# Load video
video_path = "/content/cylinder.mp4"  # replace with uploaded filename
cap = cv2.VideoCapture(video_path)

# Output video writer
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("counted_output.mp4", fourcc, fps, (width, height))

total_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run inference
    results = model(frame)[0]

    # Draw detections
    boxes = results.boxes
    count = len(boxes)
    total_count += count

    annotated_frame = results.plot()
    cv2.putText(annotated_frame, f"Cylinders in Frame: {count}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    out.write(annotated_frame)

cap.release()
out.release()

print(f"✅ Total cylinders detected across video: {total_count}")
```
