```python
# cifar10_grayscale_cnn.py
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns  # optional (pip install seaborn)

# -------------------------
# Repro & info
# -------------------------
np.random.seed(42)
tf.random.set_seed(42)
print("TensorFlow version:", tf.__version__)

# -------------------------
# 1) Read data & convert to grayscale
# -------------------------
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
y_train = y_train.squeeze()
y_test = y_test.squeeze()
class_names = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']

# Convert numpy arrays to float32 and to range [0,1]
x_train = x_train.astype('float32') / 255.0
x_test  = x_test.astype('float32')  / 255.0

# Convert to grayscale: use TensorFlow op but applied to numpy via tf.image.rgb_to_grayscale
x_train_gray = tf.image.rgb_to_grayscale(x_train).numpy()  # shape (N,32,32,1)
x_test_gray  = tf.image.rgb_to_grayscale(x_test).numpy()

print("Shapes (grayscale):", x_train_gray.shape, x_test_gray.shape)

# -------------------------
# 2) Visualize samples (grayscale)
# -------------------------
def show_gray_samples(images, labels, class_names, rows=3, cols=5):
    plt.figure(figsize=(cols*2, rows*2))
    idxs = np.random.choice(len(images), rows*cols, replace=False)
    for i, idx in enumerate(idxs):
        plt.subplot(rows, cols, i+1)
        plt.imshow(images[idx].squeeze(), cmap='gray')
        plt.title(class_names[labels[idx]])
        plt.axis('off')
    plt.tight_layout()
    plt.show()

print("Sample grayscale images:")
show_gray_samples(x_train_gray, y_train, class_names, rows=3, cols=5)

# -------------------------
# 3) Prepare tf.data pipeline (with TF-image augmentation)
# -------------------------
BATCH_SIZE = 64
AUTOTUNE = tf.data.AUTOTUNE
IMG_SIZE = (32, 32)  # CIFAR native

def normalize_img(image, label):
    # image already in [0,1], but ensure dtype
    image = tf.cast(image, tf.float32)
    return image, label

# augmentation using pure tf.image ops (works robustly)
def tf_augment(image, label):
    # image is [H,W,1] float32 in [0,1]
    # random horizontal flip (works fine for grayscale)
    image = tf.image.random_flip_left_right(image)

    # random brightness & contrast
    image = tf.image.random_brightness(image, max_delta=0.08)
    image = tf.image.random_contrast(image, lower=0.9, upper=1.1)

    # random shift via pad + crop
    h = IMG_SIZE[0]
    w = IMG_SIZE[1]
    padded = tf.image.resize_with_crop_or_pad(image, h + 4, w + 4)
    image = tf.image.random_crop(padded, size=[h, w, 1])

    # random zoom (scale then center crop/pad)
    scale = tf.random.uniform([], 0.92, 1.08)
    new_h = tf.cast(tf.cast(h, tf.float32) * scale, tf.int32)
    new_w = tf.cast(tf.cast(w, tf.float32) * scale, tf.int32)
    image = tf.image.resize(image, (new_h, new_w))
    image = tf.image.resize_with_crop_or_pad(image, h, w)

    # clip
    image = tf.clip_by_value(image, 0.0, 1.0)
    return image, label

# Create val split
VAL_SPLIT = 0.1
num_val = int(len(x_train_gray) * VAL_SPLIT)
x_val = x_train_gray[:num_val]
y_val = y_train[:num_val]
x_train_sub = x_train_gray[num_val:]
y_train_sub = y_train[num_val:]

# Training dataset (with augmentation)
train_ds = tf.data.Dataset.from_tensor_slices((x_train_sub, y_train_sub))
train_ds = train_ds.shuffle(10000, seed=42)
train_ds = train_ds.map(normalize_img, num_parallel_calls=AUTOTUNE)
train_ds = train_ds.map(tf_augment, num_parallel_calls=AUTOTUNE)   # augmentation here uses TF ops
train_ds = train_ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)

# Validation dataset (no augmentation)
val_ds = tf.data.Dataset.from_tensor_slices((x_val, y_val))
val_ds = val_ds.map(normalize_img, num_parallel_calls=AUTOTUNE)
val_ds = val_ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)

# Test dataset (no augmentation)
test_ds = tf.data.Dataset.from_tensor_slices((x_test_gray, y_test))
test_ds = test_ds.map(normalize_img, num_parallel_calls=AUTOTUNE)
test_ds = test_ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)

print("Dataset sizes -> train:", len(x_train_sub), "val:", len(x_val), "test:", len(x_test_gray))

# -------------------------
# 4) Define a CNN model (input shape = 32x32x1)
# -------------------------
def make_gray_cnn(input_shape=(32,32,1), num_classes=10):
    inputs = layers.Input(shape=input_shape)

    x = layers.Conv2D(32, (3,3), padding='same', activation='relu')(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(32, (3,3), padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.2)(x)

    x = layers.Conv2D(64, (3,3), padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(64, (3,3), padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(128, (3,3), padding='same', activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.4)(x)

    x = layers.Flatten()(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)

    outputs = layers.Dense(num_classes, activation='softmax')(x)
    model = models.Model(inputs=inputs, outputs=outputs, name='cifar10_gray_cnn')
    return model

model = make_gray_cnn(input_shape=(32,32,1), num_classes=10)
model.summary()

# -------------------------
# 5) Hyperparameters & optimizer
# -------------------------
learning_rate = 0.001
epochs = 30
optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)

model.compile(optimizer=optimizer,
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1),
    ModelCheckpoint('best_cifar10_gray_cnn.h5', monitor='val_loss', save_best_only=True, verbose=1),
]

# -------------------------
# 6) Train & visualize history
# -------------------------
history = model.fit(train_ds,
                    validation_data=val_ds,
                    epochs=epochs,
                    callbacks=callbacks,
                    verbose=1)

def plot_history(history):
    hist = history.history
    epochs_range = range(1, len(hist['loss'])+1)
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    plt.plot(epochs_range, hist['loss'], label='train_loss')
    plt.plot(epochs_range, hist['val_loss'], label='val_loss')
    plt.title('Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(epochs_range, hist['accuracy'], label='train_acc')
    plt.plot(epochs_range, hist.get('val_accuracy', []), label='val_acc')
    plt.title('Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.tight_layout()
    plt.show()

plot_history(history)

# -------------------------
# 7) Testing: evaluate + metrics
# -------------------------
test_loss, test_acc = model.evaluate(test_ds, verbose=1)
print(f"\nTest loss: {test_loss:.4f}, Test accuracy: {test_acc:.4f}")

# Predict on test set for metrics
y_pred_probs = model.predict(test_ds)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = y_test

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(9,7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix - CIFAR-10 (grayscale)')
plt.show()
```
